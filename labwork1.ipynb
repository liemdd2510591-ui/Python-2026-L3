```python
n = int(input("Enter a number? "))
if n > 1:
    is_prime = True
    for i in range(2, n ):
        if n % i == 0:
            is_prime = False
            break
    if is_prime:
        print(f"{n} is a prime number")
    else:
        print(f"{n} is a NOT prime number")
else:
    print(f"{n} is a NOT prime number")
```

```python

n = int(input("Enter a number? "))
if n > 0:
    divisor_sum = 0
    for i in range(1, n):
        if n % i == 0:
            divisor_sum += i

    if divisor_sum == n:
        print(f"{n} is a perfect number")
    else:
        print(f"{n} is a NOT perfect number")
else:
    print(f"{n} is a NOT perfect number")
```

```python
def remove_dollar_sign(s):
    return s.replace("$", "")
print(remove_dollar_sign("100$"))
```

```python
def extract_even(l):
    return [item for item in l if item % 2 == 0]
print(extract_even([1, 4, 5, -1, 10]))
```

```python
def calculate_factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
print(calculate_factorial(5))
```

```python
def get_divisors(n):
    divisors = []
    for i in range(1, abs(n) + 1):
        if n % i == 0:
            divisors.append(i)
    return divisors
print(get_divisors(12))
```

```python
import math
def compute_distance(p1, p2):
    return math.sqrt((p2[0] - p1[0])**2 + (p2[1] - p1[1])**2)
point1 = (1, 2)
point2 = (4, 6)
print(compute_distance(point1, point2))
```

```python
def print_pattern(m, n):
    for i in range(m):
        if i == 0 or i == m - 1:
            print("* " * n)
        else:
            if n > 1:
                print("* " + "  " * (n - 2) + "*")
            else:
                print("*")
print_pattern(4, 5)
```
