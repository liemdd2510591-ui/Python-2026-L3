```python
students, courses, marks = [], [], {}

# --- INPUT FUNCTIONS ---
def get_input():
    for _ in range(int(input("Enter number of students: "))):
        students.append({'id': input("Student ID: "), 'name': input("Name: "), 'dob': input("DoB: ")})

    for _ in range(int(input("Enter number of courses: "))):
        courses.append({'id': input("Course ID: "), 'name': input("Course Name: ")})

def add_marks():
    cid = input("Enter Course ID to input marks: ")
    marks[cid] = {s['id']: float(input(f"Mark for {s['name']} ({s['id']}): ")) for s in students}

# --- LISTING FUNCTIONS ---
def list_courses():
    print("\n--- Courses ---", *[f"{c['id']}: {c['name']}" for c in courses], sep="\n")

def list_students():
    print("\n--- Students ---", *[f"{s['id']}: {s['name']} (DoB: {s['dob']})" for s in students], sep="\n")

def show_marks():
    cid = input("Enter Course ID to view marks: ")
    print(f"\n--- Marks for Course {cid} ---")
    for sid, mark in marks.get(cid, {}).items():
        print(f"Student ID: {sid} | Mark: {mark}")

# --- MAIN RUNNER ---
get_input()
add_marks()
list_courses()
list_students()
show_marks()
```
