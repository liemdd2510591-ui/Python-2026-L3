```python
import math
import numpy as np

def floor_to_1_decimal(score):
    """Rounds down a score to 1 decimal place using math.floor."""
    return math.floor(score * 10) / 10

class Student:
    def __init__(self, student_id, name):
        self.id = student_id
        self.name = name
        self.marks = []
        self.credits = []
        self.gpa = 0.0

    def add_mark(self, mark, credit):
        rounded_mark = floor_to_1_decimal(mark)
        self.marks.append(rounded_mark)
        self.credits.append(credit)

    def calculate_gpa(self):
        """Calculates GPA as a weighted average using NumPy arrays."""
        if not self.marks or sum(self.credits) == 0:
            self.gpa = 0.0
            return self.gpa

        marks_array = np.array(self.marks)
        credits_array = np.array(self.credits)

        weighted_sum = np.sum(marks_array * credits_array)
        total_credits = np.sum(credits_array)

        self.gpa = weighted_sum / total_credits
        return self.gpa


def sort_students_by_gpa(students_list):
    """Sorts a list of Student objects by GPA descending using NumPy array argsort."""
    gpas = np.array([student.calculate_gpa() for student in students_list])

    sorted_indices = np.argsort(gpas)[::-1]
    return [students_list[i] for i in sorted_indices]

s1 = Student("S001", "Alice")
s1.add_mark(8.7, 3)
s1.add_mark(9.1, 2)

s2 = Student("S002", "Bob")
s2.add_mark(6.5, 3)
s2.add_mark(7.8, 4)

s3 = Student("S003", "Charlie")
s3.add_mark(9.8, 3)
s3.add_mark(8.9, 3)

students = [s1, s2, s3]

sorted_students = sort_students_by_gpa(students)

print(f"{'ID':<10} {'Name':<15} {'GPA':<10}")
print("-" * 35)
for s in sorted_students:
    print(f"{s.id:<10} {s.name:<15} {s.gpa:.2f}")
```
