```python
import os
import zipfile
import shutil

DATA_ARCHIVE = 'students.dat'
STUDENTS_FILE = 'students.txt'
COURSES_FILE = 'courses.txt'
MARKS_FILE = 'marks.txt'

students = []
courses = []
marks = {}

def decompress_and_load():
    """Checks if students.dat exists, decompresses it, and loads data into memory."""
    global students, courses, marks
    if not os.path.exists(DATA_ARCHIVE):
        print("No existing 'students.dat' found. Starting fresh.\n")
        return
    print(f"Found '{DATA_ARCHIVE}'. Decompressing files...")
    with zipfile.ZipFile(DATA_ARCHIVE, 'r') as zip_ref:
        zip_ref.extractall('.')

    if os.path.exists(STUDENTS_FILE):
        with open(STUDENTS_FILE, 'r') as f:
            for line in f:
                parts = line.strip().split(',')
                if len(parts) == 3:
                    students.append({'id': parts[0], 'name': parts[1], 'dob': parts[2]})

    if os.path.exists(COURSES_FILE):
        with open(COURSES_FILE, 'r') as f:
            for line in f:
                parts = line.strip().split(',')
                if len(parts) == 2:
                    courses.append({'id': parts[0], 'name': parts[1]})

    if os.path.exists(MARKS_FILE):
        with open(MARKS_FILE, 'r') as f:
            for line in f:
                parts = line.strip().split(',')
                if len(parts) == 3:
                    marks[(parts[0], parts[1])] = float(parts[2])
    print("Data successfully loaded from archive!\n")

def save_and_compress():
    """Writes memory data to .txt files and compresses them into students.dat."""
    with open(STUDENTS_FILE, 'w') as f:
        for s in students:
            f.write(f"{s['id']},{s['name']},{s['dob']}\n")

    with open(COURSES_FILE, 'w') as f:
        for c in courses:
            f.write(f"{c['id']},{c['name']}\n")

    with open(MARKS_FILE, 'w') as f:
        for (c_id, s_id), mark in marks.items():
            f.write(f"{c_id},{s_id},{mark}\n")

    print("\nCompressing data files into 'students.dat'...")
    with zipfile.ZipFile(DATA_ARCHIVE, 'w', zipfile.ZIP_DEFLATED) as zip_ref:
        for file_path in [STUDENTS_FILE, COURSES_FILE, MARKS_FILE]:
            if os.path.exists(file_path):
                zip_ref.write(file_path
                              , os.path.basename(file_path))
                os.remove(file_path)
    print(f"All files compressed and saved to '{DATA_ARCHIVE}'. Goodbye!")

def input_students():
    num_students = int(input("Enter number of students to add: "))
    for _ in range(num_students):
        s_id = input("  Student ID: ")
        name = input("  Student Name: ")
        dob = input("  Date of Birth (DD/MM/YYYY): ")
        students.append({'id': s_id, 'name': name, 'dob': dob})

def input_courses():
    num_courses = int(input("Enter number of courses to add: "))
    for _ in range(num_courses):
        c_id = input("  Course ID: ")
        c_name = input("  Course Name: ")
        courses.append({'id': c_id, 'name': c_name})

def input_marks():
    if not courses or not students:
        print("Please enter courses and students first!")
        return
    print("\nAvailable Courses:")
    for c in courses:
        print(f" - [{c['id']}] {c['name']}")

    target_course_id = input("Enter Course ID to input marks for: ")
    course_exists = any(c['id'] == target_course_id for c in courses)
    if not course_exists:
        print("Course ID not found.")
        return
    for s in students:
        mark = float(input(f"Enter mark for student {s['name']} (ID: {s['id']}): "))
        marks[(target_course_id, s['id'])] = mark

def show_students():
    print("\n--- Student List ---")
    if not students:
        print("No students found.")
        return
    for s in students:
        print(f"ID: {s['id']} | Name: {s['name']} | DoB: {s['dob']}")

def show_courses():
    print("\n--- Course List ---")
    if not courses:
        print("No courses found.")
        return
    for c in courses:
        print(f"ID: {c['id']} | Name: {c['name']}")

def show_marks():
    print("\n--- Marks ---")
    if not marks:
        print("No marks recorded.")
        return
    for (c_id, s_id), mark in marks.items():
        print(f"Course ID: {c_id} | Student ID: {s_id} | Mark: {mark}")

def main():
    decompress_and_load()
    while True:
        print("\n===== STUDENT MANAGEMENT MENU =====")
        print("1. Add Students")
        print("2. Add Courses")
        print("3. Input Marks")
        print("4. Show Students")
        print("5. Show Courses")
        print("6. Show Marks")
        print("7. Exit & Save (Compress to students.dat)")
        choice = input("Select an option (1-7): ")

        if choice == '1':
            input_students()
        elif choice == '2':
            input_courses()
        elif choice == '3':
            input_marks()
        elif choice == '4':
            show_students()
        elif choice == '5':
            show_courses()
        elif choice == '6':
            show_marks()
        elif choice == '7':
            save_and_compress()
            break
        else:
            print("Invalid selection, please try again.")

if __name__ == '__main__':
    main()
```
