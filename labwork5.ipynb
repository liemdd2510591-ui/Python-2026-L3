```python
import pandas as pd

students_df = pd.read_csv('students.csv')
scores_df = pd.read_csv('scores.csv')

print("\n Missing Values in Students and Scores")
print(students_df.isnull().sum())
print(scores_df.isnull().sum())

students_clean = students_df.dropna(subset=['student_id'])
scores_clean = scores_df.dropna(subset=['student_id'])

score_cols = ['python', 'math', 'database']
for col in score_cols:
    if col in scores_clean.columns:
        scores_clean[col] = scores_clean[col].fillna(scores_clean[col].mean())

merged_df = pd.merge(students_clean, scores_clean, on='student_id', how='inner')
print("\n Merged Dataset Preview")
print(merged_df.head())

merged_df['average_score'] = merged_df[['python', 'math', 'database']].mean(axis=1)
print("\n Dataset with Average Scores")
print(merged_df[['student_id', 'name', 'average_score']])

top_5_students = merged_df[['student_id', 'name', 'average_score']].sort_values(by='average_score', ascending=False).head(5)
print("\n Top 5 Students")
print(top_5_students)

avg_by_major = merged_df.groupby('major')['average_score'].mean().reset_index()
print("\n Average Score by Major")
print(avg_by_major)
```
