```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(-5, 5, 400)
y = 2 * x + np.sin(x)**2 + 3
plt.figure(figsize=(8, 6))
plt.plot(x, y, label=r'$y = 2x + \sin^2(x) + 3$')
plt.title("hello, po!")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.savefig("po.pdf")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('scores.csv')

plt.figure(figsize=(8, 6))
plt.scatter(df.index, df['math'], label='Math', alpha=0.7)
plt.scatter(df.index, df['python'], label='Python', alpha=0.7)
plt.scatter(df.index, df['database'], label='Database', alpha=0.7)

plt.title('Student Scores')
plt.xlabel('Student Index')
plt.ylabel('Scores')
plt.legend()
plt.grid(True)
plt.savefig('scattered.png')
plt.show()

fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)

axes[0].hist(df['math'], color='skyblue', edgecolor='black')
axes[0].set_title('Math Marks')
axes[0].set_xlabel('Score')
axes[0].set_ylabel('Frequency')

axes[1].hist(df['python'], color='lightgreen', edgecolor='black')
axes[1].set_title('Python Marks')
axes[1].set_xlabel('Score')

axes[2].hist(df['database'], color='salmon', edgecolor='black')
axes[2].set_title('Database Marks')
axes[2].set_xlabel('Score')

plt.tight_layout()
plt.savefig('histo.png')
plt.show()

plt.figure(figsize=(8, 6))
df[['math', 'python', 'database']].boxplot()

plt.title('Box Plot')
plt.ylabel('Scores')
plt.grid(True)
plt.savefig('box.png')
plt.show()
```
